```sql
-- =============================================================
-- Unity Catalog setup for the Olist lakehouse
-- Run once per environment. Safe to re-run (IF NOT EXISTS).
--
-- Layering:
--   landing -> bronze -> silver -> gold -> mart_<area>_<report>
--   * mart_* schemas are created by dbt (one schema per report/consumer)
-- =============================================================

-- Catalog = environment (olist_dev / olist_prod)
CREATE CATALOG IF NOT EXISTS olist_dev
  COMMENT 'Olist lakehouse – development environment';

-- Schemas = layers
CREATE SCHEMA IF NOT EXISTS olist_dev.landing
  COMMENT 'Raw source files as delivered (volumes)';

CREATE SCHEMA IF NOT EXISTS olist_dev.bronze
  COMMENT 'Raw data 1:1 from source + ingestion metadata (Auto Loader, API)';

CREATE SCHEMA IF NOT EXISTS olist_dev.silver
  COMMENT 'Cleaned, typed, deduplicated data (dbt staging / intermediate)';

CREATE SCHEMA IF NOT EXISTS olist_dev.gold
  COMMENT 'Core star schema: conformed dimensions and facts, single source of truth (dbt)';

-- Volumes
CREATE VOLUME IF NOT EXISTS olist_dev.landing.raw
  COMMENT 'Landing zone for source files (one folder per table)';

CREATE VOLUME IF NOT EXISTS olist_dev.bronze.checkpoints
  COMMENT 'Auto Loader checkpoints and schema tracking';
```
