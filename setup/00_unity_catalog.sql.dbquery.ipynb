```sql
-- Unity Catalog setup for the Olist lakehouse (run once per environment)

CREATE CATALOG IF NOT EXISTS olist_dev
  COMMENT 'Olist lakehouse – development environment';

CREATE SCHEMA IF NOT EXISTS olist_dev.landing    COMMENT 'Raw source files (volume)';
CREATE SCHEMA IF NOT EXISTS olist_dev.bronze     COMMENT 'Raw data 1:1 + ingestion metadata';
CREATE SCHEMA IF NOT EXISTS olist_dev.silver     COMMENT 'dbt staging / intermediate';
CREATE SCHEMA IF NOT EXISTS olist_dev.gold       COMMENT 'dbt star schema (dim/fct)';
CREATE SCHEMA IF NOT EXISTS olist_dev.mart_sales COMMENT 'dbt mart – sales';

CREATE VOLUME IF NOT EXISTS olist_dev.landing.raw
  COMMENT 'Landing zone for source files';

CREATE VOLUME IF NOT EXISTS olist_dev.bronze.checkpoints
  COMMENT 'Auto Loader checkpoints and schema tracking';
```
